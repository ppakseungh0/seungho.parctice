# S6E11 | Two Ladders, One Step: What Actually Moved LB

**Our earlier versions** (all in the version history): v15 0.95333, v13 0.95330, v12 0.95300, v11 0.95291, v10 0.95288, v9 0.95239. Our best file upload before this version: 0.95336.

This version adds one step to the best public file. It takes amanatar's Strike 10 file (0.95340), which climbs from Moran Z's Strike 9 (0.95330), and adds the language-model step that took our own ladder from 0.95333 to 0.95336. Its public score is shown in the box above. Almost all of the score is the work of Moran Z and amanatar; read **How to read the gaps** before you use it as a final pick.

## What it does

1. Loads three public files and checks each against its SHA-256:
   - `submission_095340.csv` (`A`): Strike 10 from [S6E11 Strike 10 Podium Apex LB 0.95340](https://www.kaggle.com/datasets/amanatar/s6e11-strike10-podium-apex-095340) by amanatar (CC0).
   - `submission_095330.csv` (`F`): Strike 9 from [S6E11 Strike9 LB 0.95330](https://www.kaggle.com/datasets/moranzavdi/s6e11-strike9-lb-095330) by Moran Z (CC0).
   - `lm_test.npy` (`LM`): a GPT-2 that reads each row label first and scores it with Bayes' rule, from [S6E11 target-first LM predictions](https://www.kaggle.com/datasets/megayak/s6e11-target-first-lm) by megayak (CC0).
2. Turns each into logit ranks, `logit((rank - 0.5) / n)`. AUC only sees order, so this puts all three on one scale.
3. Adds the language-model step to Strike 10, measured from their common base: `score = A + 0.12 (LM - F)`.
4. Ranks the result, writes `submission.csv` and checks its SHA-256.

## Why this step: our own ladder

**Round 1, from Strike 9 `F`** (v15). `MLP` is goodpjw2008's `gbn_bag5_w512` and `GB` is `gb_bag5`, both from the [S6E11 OOF Bank](https://www.kaggle.com/datasets/goodpjw2008/s6e11-oof-bank).

| Probe | Public LB |
|---|---|
| F unchanged (v13) | 0.95330 |
| 0.90 F + 0.10 MLP | 0.95326 |
| F + 0.05 / 0.10 / 0.15 (F - MLP) | 0.95332 / 0.95332 / 0.95332 |
| **F + 0.10 (F - MLP) + 0.05 (F - GB) = B (v15)** | **0.95333** |

**Round 2, from `B`**:

| Probe | Public LB |
|---|---|
| 0.97 / 0.95 / 0.92 B + 0.03 / 0.05 / 0.08 LM | 0.95335 / 0.95335 / 0.95335 |
| **0.88 B + 0.12 LM** | **0.95336** |
| 0.97 B + 0.03 masayakawamata GPT-2 | 0.95332 |
| 0.97 B + 0.03 our own target-first transformer | 0.95332 |
| B + 0.05 (B - gbn_bag5_w128) | 0.95332 |
| B + 0.05 (B - ebm_dig_bag) | 0.95331 |

Collinear tree and MLP bags dilute the file; megayak's language model is the only member that gains when added, and the gain is a wide plateau from 3% to 12%.

## Why it should stack on Strike 10

We decomposed each file of amanatar's public ladder into moves from Strike 9 (least squares on logit ranks):

| File | Public LB | weight on MLP | weight on GB | weight on LM | share not explained by these |
|---|---|---|---|---|---|
| amanatar Strike 10 step | 0.95334 | -0.028 | -0.020 | +0.021 | 0.89 |
| amanatar Strike 10 step | 0.95337 | -0.023 | -0.017 | -0.008 | 0.90 |
| amanatar Strike 10 step | 0.95339 | -0.036 | -0.026 | -0.003 | 0.88 |
| **amanatar Strike 10** | **0.95340** | **-0.047** | **-0.034** | **-0.001** | **0.87** |
| our 0.88 B + 0.12 LM | 0.95336 | -0.088 | -0.044 | +0.120 | 0.00 |

- His climb also steps away from the same two members as ours, but 87% of it is moves we cannot see.
- **It carries almost no language model** (weight -0.001 against our +0.120). His whole move from Strike 9 correlates only -0.13 with our language-model step, so the two should add rather than overlap.

## CV evidence: these steps do not hold up

Strike 9 and Strike 10 have no OOF, so we test our moves on the nested OOF of goodpjw2008's Part 12 wide LM stack (AUC 0.952565), the closest public stand-in:

| Move on the wide-stack OOF | CV change |
|---|---|
| 0.10 away from MLP + 0.05 away from GB (v15) | -7.4e-5 |
| + 5% megayak LM (all 5 folds) | -1.5e-5 |
| + 10% megayak LM (fold 0) | -14.7e-5 |

Every move that wins on the public LB loses on CV. The stack already contains this LM, so adding it again only pays on the public 20%.

## What did not work

We trained our own target-first language model to add a new kind of error: a small transformer that reads each row label first, one token per field plus the GPT-2 tokens of its value (about 66 tokens per row), scored with Bayes' rule over 8 field orders. On fold 0, after 45 minutes on a T4, it reached AUC 0.94685 (5M parameters, 12.5 epochs); a 15M version reached 0.94610. It correlates 0.976 with the public LMs and loses CV at every weight (-3.9e-5 at 3%, -15.3e-5 at 10%). The full GPT-2 recipe from the S6E10 winners does not fit a T4 for medium size and needs about 130 GPU hours for 5 folds of small.

## How to read the gaps

- The public LB is about 20% of the test set, roughly 60k rows. 0.00001 is noise and 0.00003 is barely more.
- Strike 9 and Strike 10 contain LB-gradient steps by their authors, and this notebook adds one more. All of them can fade on the private 80%.
- **For your two final picks, treat this as the best-public pick and pair it with a CV-backed one**, such as our v12 (0.95300).
- The next real gain is a model with different errors, checked on OOF first, not more blending.

```python
from pathlib import Path
import hashlib
import numpy as np
import pandas as pd
from scipy.stats import rankdata
from scipy.special import logit

# Public inputs, checked byte for byte
EXPECTED = {
    'submission_095340.csv': '12b525b84f29e32d6511e9d6105f6dc5eac79ae41b10140eed3353ea57afc159',  # Strike 10 by amanatar
    'submission_095330.csv': 'f4a7a4f8636c99fabd82945c7d81956c5559c609dcb55cde6d4e03f80b0ea105',  # Strike 9 by Moran Z
    'lm_test.npy': '3168779be644a530c6bdc4c141cae168bfc2feb01ac21090c917cc978a7687d3',  # megayak target-first LM
}
OUT_SHA = 'b836b793b46621dcda1568905c1a0dbeb2bb8c23c86cbc2980cb5ddd024f1384'  # this version's submission
INPUT = Path('/root/.cache/kagglehub')
OUT = Path('/kaggle/working/submission.csv')

def locate(name):
    hits = sorted(INPUT.rglob(name))
    assert hits, name
    return hits[0]

def sha256(path):
    return hashlib.sha256(Path(path).read_bytes()).hexdigest()

def logit_rank(v):
    return logit((rankdata(v) - 0.5) / len(v))

paths = {name: locate(name) for name in EXPECTED}
for name, path in paths.items():
    assert sha256(path) == EXPECTED[name], (name, sha256(path))

sample = pd.read_csv(locate('sample_submission.csv'))
strike10 = pd.read_csv(paths['submission_095340.csv'])
strike9 = pd.read_csv(paths['submission_095330.csv'])
for sub in (strike10, strike9):
    assert (sub['id'].to_numpy() == sample['id'].to_numpy()).all()

A = logit_rank(strike10['is_fraud'].to_numpy())
F = logit_rank(strike9['is_fraud'].to_numpy())
LM = logit_rank(np.load(paths['lm_test.npy']))
assert len(A) == len(F) == len(LM) == len(sample)

# Strike 10 plus the language-model step, measured from the common base Strike 9
score = A + 0.12 * (LM - F)
print('correlation with Strike 10:', round(float(np.corrcoef(score, A)[0, 1]), 6))

# Ensure the directory exists before saving
OUT.parent.mkdir(parents=True, exist_ok=True)

sample.assign(is_fraud=(rankdata(score) - 0.5) / len(score)).to_csv(OUT, index=False)
assert sha256(OUT) == OUT_SHA, sha256(OUT)
print('wrote', OUT, len(sample), 'rows, sha', OUT_SHA[:16])
```
